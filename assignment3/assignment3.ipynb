```python
# import statements

import requests as re
import json
import csv
import time
```

```python
LINK = 'https://www.loc.gov/free-to-use/libraries/'
```

```python
params = {
    'fo':'json'
}
```

```python
libraries = re.get(LINK, params=params)
libraries = libraries.json()
```

```python
with open('libraries.json', 'w', encoding='utf-8') as file:
    json.dump(libraries, file, indent=4)
```

```python
# finding 'items' key

len(libraries['content']['components'][0]['masonry_gallery'][0]['items'])

lib_items = libraries['content']['components'][0]['masonry_gallery'][0]['items']
```

```python
# write lib_items to csv

with open('ftu-libraries-set-list.csv', 'w', newline='') as csvfile:
    fieldnames = ['image', 'link', 'title']
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(lib_items)
```

```python
baseURL = 'https://www.loc.gov'
with open('ftu-libraries-set-list.csv', mode='r', newline='', encoding='utf-8') as csvfile:
    reader = csv.DictReader(csvfile)
    for index, row in enumerate(reader):
        resource = row['link']
        link = baseURL + resource
        item = re.get(link, params=params)

        if item.status_code == 200:
            item_data = item.json()
            with open(f'item{index}.json', 'w', encoding='utf-8') as file:
                json.dump(item_data, file, indent=4)

        else:
            print(f'failed to fetch item with index of {index}: Status {item.status_code}')

        time.sleep(1)
```
